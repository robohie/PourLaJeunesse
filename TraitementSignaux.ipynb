<h2>Introduction aux signaux et systèmes</h2>

<p>On appelle <b>signal</b> toute quantité qui sert de support d'information que nous pouvons communiquer. Nous pouvons citer : la tension électrique, le courant électrique, la pression accoustique ou les ondes sismiques.</p>
<img src="signaux.png">

<p>Un <b>système</b> peut être défini comme un dispositif qui traite des signaux. Il accepte un ou plusieurs signaux en entrée et ressort, après traitement, un ou plusieurs signaux en sortie. Une bonne façon de réfléchir est de considérer le monde qui nous entoure comme un gros système contenant d'autres systèmes qui se communiquent ; on ramène ainsi les problèmes de la vie vers un modèle orienté système.</p>
<img src="systèmes.png">

<p>Le travail d'un ingénieur peut se résumer d'une part par la transformation des problèmes du monde "réel" vers le monde mathématique(la modélisation) à la recherche des solutions optimales, d'autre part par l'interprétation des solutions trouvées dans ce monde fictif afin de répondre resoudre les problèmes d'origine. Ce monde mathématique est fait des équations, des fonctions, des espaces vectoriels... Nous devons pour ce faire représenter les signaux ainsi que les systèmes dans cette réalité mathématique. Les signaux sont représentés par des fonctions, les systèmes le sont par des équations ou des algorithmes.</p>

<h3>Type des signaux</h3>
<p>Sachant qu'une fonction est défini comme une correspondance entre un ensemble de départ et un ensemble d'arrivé, la classification des signaux peut se faire en fonction de ces ensembles. Quand nous considérons l'ensemble de départ, nous avons :</p>
* Les signaux à temps continu : représentés par des fonctions continues dans le temps. Par exemple, si nous considérons le signal sinusoïdal $x(t)=\sin(x)$ pour tout réel t, il s'agit d'une fonction continue et donc d'un signal continu dans le temps.
<img src="signal_temps_continu.png">
* Les signaux à temps discret : représentés par des fonctions à temps discontinu i.e. la valeur du signal n'est connu que pour un ensemble de départ constitué des entiers.
<img src="signal_temps_discret">

<p>Si nous considérons par contre l'ensemble d'arrivé, les signaux peuvent être classés comme ci-dessous :</p>
<p>Les signaux digitaux, qui sont des signaux à temps discret tel que l'ensemble d'arrivé est un sous-ensemble de l'ensemble des entiers. Grâce à la <b>quantification</b>, nous pouvons passer d'un signal à temps continu à un signal à temps discret, en suivant le <b>théotème d'échantillonnage de Claude Shannon</b>.</p>
<p>Les signaux numériques, une sous-classe des signaux digitaux dont l'ensemble d'arrivé est formé de $0$ et $1$.</p>

<h3>Energie et puissance d'un signal</h3>
<p>On appelle <b>energie d'un signal</b> dans l'intervalle de temps $[t_1, t_2]$(respectivement $[n_1, n_2]$) la grandeur : $E(x(t))=\int_{t_1}^{t_2}{|x(t)|^2}dt$(respectivement $E(x[n])=\sum \limits_{n=n_1}^{n_2}{|x[n]|^2}dt$)</p>
<p>Un signal est appelé <b>signal énergie</b>, son énergie totale est finie(pour $t$ qui varie de moins l'infini à plus l'infini)</p>

<p><b>La puissance d'un signal</b> à temps continu ou à temps discret est définie respectivement comme : $P(x(t))=\lim \limits_{T \to +\infty} \frac{E(x(t))}{2T}$ et $P[x[n]]=\lim \limits_{N \to +\infty} \frac{E[x[n]]}{2N+1}$</p>
<p>Un signal est appelé <b>signal puissance</b> si sa puissance est différente de zéro et finie.</p>

<h3>Opérations sur a variable temporelle d'un signal</h3>

```python
# Décalage temporel.
# Nous partons d'un signal x(t) vers un signal x(t-a)
import numpy as np
import matplotlib.pyplot as plt
from signaux_systems_modules.operations import *

t = np.linspace(-2 * np.pi, 2 * np.pi, 400)  # le variable temps
x = lambda temps: np.sin(temps) + 2 * np.cos(temps)**2 # nous considérons un signal sinus
a:float = 0.5 # le pas de décalage, positive à droite

# décalage(time shift) : x(t-a)
# Voir le module operations pour plus d'info sur la fonction shift
plt.plot(t, x(t), label="x(t)")
plt.plot(t, shift_time(x, a, t), label=f"x(t-{a})")
plt.grid()
plt.legend()
plt.xlabel("t[s]")
plt.ylabel("x[SI]")
plt.title("Décalage temporel (time shift)")
```

```python
# Inversion du temps(time reverse).
# Nous de x(t) vers x(-t)

# Voir le module operations pour plus d'info sur la fonction scale_time
plt.plot(t, x(t), label="x(t)")
plt.plot(t, scale_time(x, -1, t), label="$x(-t)$")
plt.grid()
plt.legend()
plt.xlabel("t[s]")
plt.ylabel("x[SI]")
plt.title("inversion temporelle (time reversal)")
```

```python
# Dilatation du temps (time scale)
# Ressemble à un zoom sur une image

a:float = 2

plt.plot(t, x(t), label="x(t)")
plt.plot(t, scale_time(x, a, t), label=f"$x({a}.t)$")
plt.grid()
plt.legend()
plt.xlabel("t[s]")
plt.ylabel("x[SI]")
plt.title("dilatation temporelle (time scale)")
```

```python
# Décalage et dilatation combinés.
# x(t) vers x(at-b)

a:float = 2
b:float = 1

plt.plot(t, x(t), label="x(t)")
plt.plot(t, shift_scale_time(x, a, b, t), label=f"$x({a}.t-{b})$")
plt.grid()
plt.legend()
plt.xlabel("t[s]")
plt.ylabel("x[SI]")
plt.title("dilatation et décalage temporelle (time scale)")
```
